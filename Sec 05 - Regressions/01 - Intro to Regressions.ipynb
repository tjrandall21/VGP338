# OLS for single feature

```python
# (xi - xbar) * (yi - ybar)
# --------------------------  = b1
#    ( xi - xbar ) ^ 2

# b0 = ybar - b1*xbar
```

```python
class SimpleLinearRegression:
    def __init__(self):
        self.beta_0 = None
        self.beta_1 = None

    def fit(self, X, y): # train the model or use some mathemtical formula

        X = X.flatten()
        y = y.flatten()

        x_mean = X.mean()
        y_mean = y.mean()

        numerator = np.sum ((X - x_mean) * (y - y_mean))
        denominator = np.sum((X - x_mean) ** 2)
        self.beta_1 = numerator / denominator
        self.beta_0 = y_mean - self.beta_1 * x_mean


    def predit(self, X): # use it once on some data - forward pass
        X = X.flatten()
        prediction = self.beta_0 + self.beta_1 * X
        return prediction
```

```python
model = SimpleLinearRegression()
```

```python
model.fit (house_size, house_price)
```

```python
print (f"beta_0 = {model.beta_0:,.2f}")
print (f"beta_1 = {model.beta_1:.2f}")
```
