### Part 6 - Grouping, Aggregation, Windows

```python
# Pandas Grouping, Aggregation, Windows — single cell, minimal
df = pd.read_csv("bank.csv")
```

```python
# --- groupby (agg, multi-agg) ---
g = df.groupby("marital")
agg_basic = g["balance"].agg(["mean", "sum", "count"])
agg_multi = g.agg(balance_mean=("balance","mean"), qty_sum=("age","sum"))
```
