### Part 03 - Indexing, Slicing & Views

```python
# --- Slicing, step, ellipsis; row/column selection ---
row1 = X[1]            # 2nd row
col2 = X[:, 2]         # 3rd col
block = X[:2, 1:3]     # top-left 2x2 (cols 2..3)
every_other_col = X[:, ::2] #

pick = X[[0, 2], [1, 3]]     # (0,1) and (2,3)
rows = X[[2, 0]]
cols = X[:, [3, 1]]
```

```python
# --- Boolean masks & filtering ---
mask = X % 2 == 0 # X % 2
evens = X[mask]

rows_sum_gt20 = X[X.sum(axis=1) > 20]
```

```python
# --- Views vs copies; shares_memory, copy() ---
Y_view = X[:, :2]            # view when possible
Y_copy = X[:, :2].copy()
mem_view = np.shares_memory(X, Y_view)
mem_copy = np.shares_memory(X, Y_copy)
```

### Part 05 - Core Operations (UFuncs)

```python
# Elementwise ops, reductions, axis aggregations — single cell, minimal
import numpy as np

X = np.array([[1., 2., 3.],
              [4., 5., 6.]])

X + X
X - X
X * X
X / X
X**2
np.exp(X)
np.log(X)
np.sqrt(X)
np.clip(X, 2, 5)
```

```python
# --- Reductions (whole array + axis usage) ---
X = np.array([[1., 2., 3.],
              [4., 5., 6.]])

X.sum()
X.sum(axis=1, keepdims=True)
X.mean()
X.std()
X.var()
X.min()
X.max()
X.argmin()
X.argmax()
```
