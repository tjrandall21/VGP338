### Part 1 - Describing Data with Averages

Primary measures of central tendency

1. **Mode**: Appears most frequently. useful for categorical or discrete data.

2. **Median**: Middle value. Less sensitive to extremes.

3. **Mean**: Also known as the average, commonly used measure of central tendency for continuous data.
    - Mean: summing up all the values and dividing by the count.

- **Average**: representing typical or central value. Could refer to mean, median, or mode, depending on the context.


### Part 2 - Mode

**How**
- Arrange the values in a systematic order
- Count the frequency
- Find the one with the highest frequency

**Advantages & Disadvantages:**
- Mode is easy to understand
- Useful for categorical or discrete data.
- Not affected by outliers
- May not exist or may not be unique
- Only the frequencies, not the actual magnitudes
- Not good if frequencies are even

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter
```

```python
# Discreet Data - Numeric countbale
shoe_sizes = np.array([7, 8, 9, 8, 7, 10, 9, 8, 8, 7,
                      9, 8, 10, 7, 8, 9, 8, 11, 8, 9])
```

```python
# Continuous Data - Numeric within range, inc decimals
test_scores = np.array([78, 85, 92, 88, 76, 94, 87, 83, 90, 89,
                       82, 95, 88, 91, 86, 79, 93, 88, 84, 87])
```

## Key Differences

| Discrete Data | Continuous Data |
|---|---|
| Counted | Measured |
| Usually whole numbers | Can include decimals |
| Separate values | Any value within a range |
| Example: Number of siblings | Example: Height |

```python
# Categorical - Categories
ratings = ['Good', 'Excellent', 'Fair', 'Good', 'Excellent', 'Good',
          'Poor', 'Good', 'Excellent', 'Fair', 'Good', 'Good',
          'Excellent', 'Fair', 'Good', 'Excellent', 'Good', 'Fair', 'Good', 'Excellent']


# Nomial -> Categorical without order e.g., Eye color
# Ordinal -> Meaningful order with unequal gaps e.g., rating
```

```python
plt.style.use('default')
sns.set_palette("husl")

# Test scores mode
test_counter = Counter(test_scores)
test_mode_value = test_counter.most_common(1)[0][0]
test_mode_freq = test_counter.most_common(1)[0][1]

# Visualization of modes
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(15, 4))

# Test scores histogram
ax1.hist(test_scores,
         bins=8,
         alpha=0.7,
         color='skyblue',
         edgecolor='black')

ax1.axvline(test_mode_value,
            color='red',
            linestyle='--',
            linewidth=2,
            label=f'Mode: {test_mode_value}')

ax1.set_title('Test Scores Distribution')
ax1.set_xlabel('Score')
ax1.set_ylabel('Frequency')
ax1.legend()
ax1.grid(True, alpha=0.3)


# Shoe sizes mode
shoe_counter = Counter(shoe_sizes)
shoe_mode_value = shoe_counter.most_common(1)[0][0]
shoe_mode_freq = shoe_counter.most_common(1)[0][1]

# Shoe sizes bar chart
shoe_sizes_unique, shoe_counts = np.unique(shoe_sizes, return_counts=True)
bars = ax2.bar(shoe_sizes_unique,
               shoe_counts,
               alpha=0.7,
               color='lightgreen',
               edgecolor='black')

# Highlight mode
mode_index = np.where(shoe_sizes_unique == shoe_mode_value)[0][0]
bars[mode_index].set_color('red')
bars[mode_index].set_alpha(0.9)
ax2.set_title('Shoe Sizes Distribution')
ax2.set_xlabel('Shoe Size')
ax2.set_ylabel('Frequency')
ax2.grid(True, alpha=0.3)


# Customer ratings mode
rating_counter = Counter(ratings)
rating_mode_value = rating_counter.most_common(1)[0][0]
rating_mode_freq = rating_counter.most_common(1)[0][1]

# Customer ratings bar chart
rating_categories = ['Poor', 'Fair', 'Good', 'Excellent']
rating_counts = [rating_counter.get(cat, 0) for cat in rating_categories]
bars = ax3.bar(rating_categories,
               rating_counts,
               alpha=0.7,
               color='lightcoral',
               edgecolor='black')

# Highlight mode
mode_index = rating_categories.index(rating_mode_value)
bars[mode_index].set_color('red')
bars[mode_index].set_alpha(0.9)
ax3.set_title('Customer Ratings Distribution')
ax3.set_xlabel('Rating')
ax3.set_ylabel('Frequency')
ax3.tick_params(axis='x', rotation=45)
ax3.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("Red bars/lines show the mode (most frequent value) in each dataset")
```

### Part 3 - Mean

Arithmetic average is the sum of all values divided by the total number of values
Often times, it's the mean of a sample as opposed to the actual population.

<img src="https://i.ibb.co/wNwtm1s3/sample-population-mean-2.png" alt="sample-population-mean-2" border="0">

<img src="https://i.ibb.co/5h4xL4cT/sample-population-mean.png" alt="sample-population-mean" border="0">

**Sample mean**
is denoted by $\bar{x}$ (read as "x-bar"), calculated :

$\bar{x} = \frac{\sum_{i=1}^{n} x_i}{n}$

where
- $x_i$ represents each individual value in the dataset
- $n$ is the total number of values in the sample,
- $\sum$ (sigma) denotes the sum of all values.


<br><br>

**Population mean** is denoted by $\mu$ (read as "mu"), calculated:

$\mu = \frac{\sum_{i=1}^{N} x_i}{N}$

where
- $x_i$ represents each individual value in the dataset
- $N$ is the total number of values in the population
- $\sum$ (sigma) denotes the sum of all values.



**Advantages & Disadvantages**
- Accounts the actual values
- Useful for interval/ratio
- Sensitive to changes in any value in dataset
- Strongly influenced outliers
- Not appropriate for skewed distributions
- Incalculable for categorical or ordinal data

```python
salaries_normal = [45000, 48000, 52000, 47000, 49000, 51000, 46000, 50000, 48500, 49500]
salaries_with_ceo = salaries_normal + [2_500_000]  # CEO salary

mean_normal = np.mean(salaries_normal)
mean_with_ceo = np.mean(salaries_with_ceo)

print(f"  Mean: ${mean_normal:,.0f}")

print(f"\nWith CEO: {salaries_normal + [2500000]}")
print(f"  Mean: ${mean_with_ceo:,.0f} (changed by ${mean_with_ceo - mean_normal:,.0f})")
```

### Part 4 - Median

1. Arrange the values in ascending or descending order.
2. Find the middle value (or the mean of the two in the middle)

**Advantages & Disadvantages**
- Less affected by outliers
- Usable with ordinal data
- Unique and always exists
- No account the actual values of the data points, only their positions
- Not representative of datasets with large range
- Time-consuming if the data is not sorted

```python
print("MEDIAN ANALYSIS")
print("=" * 50)

# Regular employees + one CEO
salaries_normal = [45000, 48000, 52000, 47000, 49000, 51000, 46000, 50000, 48500, 49500]
salaries_with_ceo = salaries_normal + [2500000]  # CEO salary

median_normal = np.median(salaries_normal)
median_with_ceo = np.median(salaries_with_ceo)

print(f"Regular employees: ${salaries_normal}")
print(f"  Median: ${median_normal:,.0f}")

print(f"\nWith CEO: {salaries_normal + [2500000]}")
print(f"  Median: ${median_with_ceo:,.0f} (changed by ${median_with_ceo - median_normal:,.0f})")
```
