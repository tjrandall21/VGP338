### Part 1 - Fundamentals of Probability

**Probability in AI**

- **Model evaluation**: What does "95% confidence" mean? Is the result statistically significant?
- **Uncertainty quantification**: How confident should your model be in its predictions?
- **Loss functions**: Cross-entropy, KL divergence, and likelihood-based losses are all rooted in probability
- **Bayesian methods**: Probabilistic modeling, uncertainty estimation, and Bayesian neural networks
- **Generative models**: VAEs, GANs, diffusion models all require deep probability understanding

**Probability** refers to the proportion or fraction of times that a particular event is likely to occur.

Probability can be determined in several ways:
1. **Theoretical/Speculative**: Based on logical reasoning (e.g., fair coin has $P(\text{heads}) = 0.5$)
2. **Empirical/Observational**: Based on observed data (e.g., observed frequency in experiments)
3. **Mixed**: Combination of observation and theoretical models

**Key Properties:**
- Probability values range from $0$ (impossible) to $1$ (certain) [e.i. $0 \leq P(A) \leq 1$ for any event $A$]
- The sum of all probabilities for a complete set of outcomes equals $1$
> $P(\text{certain event}) = 1$ <br>
> $P(\text{impossible event}) = 0$

### Part 2 - Law of Large Numbers

The **Law of Large Numbers** states that as the number of trials increases, the observed proportion of an event converges to its true probability.

```python
import numpy as np
import matplotlib.pyplot as plt

# Demonstrate convergence to true probability with coin tosses
np.random.seed(42)

n_tosses_list = [10, 50, 100, 500, 1000, 5000, 10000]
head_proportions = []

for n_tosses in n_tosses_list:
    tosses = np.random.choice(['H', 'T'], n_tosses)
    proportion_heads = np.mean(tosses == 'H')
    head_proportions.append(proportion_heads)

# Visualize convergence
plt.figure(figsize=(12, 6))
plt.plot(n_tosses_list, head_proportions, 'bo-', linewidth=2, markersize=10,
         label='Observed Proportion')
plt.axhline(y=0.5, color='red', linestyle='--', linewidth=2,
           label='Theoretical Probability (0.5)')
plt.xscale('log')
plt.xlabel('Number of Tosses (log scale)', fontsize=12)
plt.ylabel('Proportion of Heads', fontsize=12)
plt.title('Law of Large Numbers: Observed Proportion Converges to True Probability',
         fontsize=13, fontweight='bold')
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.ylim(0.3, 0.7)

# Add annotations
for i, (n, p) in enumerate(zip(n_tosses_list, head_proportions)):
    offset = -15 if i % 2 == 0 else 15
    plt.annotate(f'{p:.3f}', xy=(n, p), xytext=(5, offset),
                textcoords='offset points', fontsize=9,
                bbox=dict(boxstyle='round,pad=0.3', facecolor='yellow', alpha=0.5))

plt.tight_layout()
plt.show()

print("📊 Probability Estimates by Sample Size:")
print("="*50)
for n, p in zip(n_tosses_list, head_proportions):
    print(f"n = {n:<5d}: P(H) = {p:.4f}")
```

### Part 3 - Addition Rule

**Mutually exclusive events** are events that cannot occur together.

**Addition Rule**: When you need to find the probability of two or more mutually exclusive events connected by the word \"or\", add their individual probabilities.

$$P(A \text{ or } B) = P(A) + P(B)$$

**Examples of Mutually Exclusive Events:**
- Rolling a 3 or a 5 on a die (can't get both in one roll)
- Blood type: being type O or type A (can't be both)
- Birth month: being born in January or February

**Important**: **General Formula for Non-Mutually Exclusive Events:**

When events A and B can occur together:
$$P(A \text{ or } B) = P(A) + P(B) - P(A \text{ and } B)$$

This formula ensures we don't double-count the cases where both events occur simultaneously.

**Mutually Exclusive Events Example**

Given the blood type figure below, can you calculate the following probabilities:

- Probability of having type O?
- Probability of having type O or A?

```python
#Example: Blood type distribution in the U.S.
blood_types = {
    'O+': 0.37, 'O-': 0.07,
    'A+': 0.36, 'A-': 0.06,
    'B+': 0.09, 'B-': 0.02,
    'AB+': 0.03, 'AB-': 0.01
}

# Visualize blood type distribution
plt.figure(figsize=(10, 6))
bars = plt.bar(blood_types.keys(), blood_types.values(), alpha=0.7, edgecolor='black')
plt.xlabel('Blood Type', fontsize=12)
plt.ylabel('Probability', fontsize=12)
plt.title('Blood Type Distribution in the U.S.', fontweight='bold', fontsize=14)
plt.grid(True, alpha=0.3, axis='y')

# Add value labels on bars
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height,
            f'{height:.2f}',
            ha='center', va='bottom', fontsize=10)

plt.tight_layout()
plt.show()
```

```python
# Calculate probabilities using addition rule
p_o_positive = blood_types['O+']
p_o_negative = blood_types['O-']
p_o_any = p_o_positive + p_o_negative  # Addition rule

p_a_any = blood_types['A+'] + blood_types['A-']

print("Blood Type Probabilities:")
print("="*50)
print(f"P(O+) = {p_o_positive:.2f}")
print(f"P(O-) = {p_o_negative:.2f}")
print(f"\nUsing Addition Rule:")
print(f"P(O) = P(O+ or O-) = P(O+) + P(O-) = {p_o_positive:.2f} + {p_o_negative:.2f} = {p_o_any:.2f}")
print(f"Similarly: \nP(A) = {p_a_any:.2f}")
print(f"\nP(O or A) = {p_o_any + p_a_any:.2f}")
```

**Non-Mutually Exclusive Events Example**

- When events **can occur together** (not mutually exclusive), we must account for the overlap.
- Consider students in a class where some are seniors and some are psychology majors – a student can be **both**.

```python
# Example: Students in a class
# Total students: 100
n_students = 100

# Define probabilities
p_senior = 0.20        # 20 students are seniors
p_psych_major = 0.70   # 70 students are psychology majors
p_both = 0.10          # 10 students are BOTH seniors AND psychology majors

# Calculate frequencies
n_senior = int(n_students * p_senior)
n_psych = int(n_students * p_psych_major)
n_both = int(n_students * p_both)




print("Class Composition (100 students):")
print("="*60)
print(f"Seniors: {n_senior} students (P = {p_senior:.2f})")
print(f"Psychology Majors: {n_psych} students (P = {p_psych_major:.2f})")
print(f"Both Senior AND Psych Major: {n_both} students (P = {p_both:.2f})")
```

**Question**: What is the probability that a randomly selected student is either a senior OR a psychology major (or both)?

**Wrong Calculation** (ignoring overlap):
$$P(\text{Senior or Psych}) = P(\text{Senior}) + P(\text{Psych}) = 0.20 + 0.70 = 0.90$$

This is **incorrect** because we've counted students who are both seniors AND psychology majors twice!

**Correct Calculation** (accounting for overlap):
$$P(\text{Senior or Psych}) = P(\text{Senior}) + P(\text{Psych}) - P(\text{Senior and Psych})$$
$$= 0.20 + 0.70 - 0.10 = 0.80$$

```python
# Visualize with a Venn diagram
import matplotlib.pyplot as plt

import matplotlib.patches as mpatches
from matplotlib.patches import Circle

fig, ax1 = plt.subplots(1, 1, figsize=(14, 6))

# Left plot: Venn diagram
ax1.set_xlim(0, 10)
ax1.set_ylim(0, 10)
ax1.set_aspect('equal')
ax1.axis('off')
ax1.set_title('Venn Diagram: Seniors and Psychology Majors', fontsize=14, fontweight='bold')

# Draw circles
circle1 = Circle((3.5, 5), 2.5, color='lightblue', alpha=0.5, ec='black', linewidth=2)
circle2 = Circle((6.5, 5), 2.5, color='lightcoral', alpha=0.5, ec='black', linewidth=2)
ax1.add_patch(circle1)
ax1.add_patch(circle2)


n_only_senior = n_senior - n_both
n_only_psych = n_psych - n_both
n_neither = n_students - n_senior - n_psych + n_both



# Add labels
ax1.text(2.5, 5, f'Only\nSeniors\n{n_only_senior}', ha='center', va='center', fontsize=12, fontweight='bold')
ax1.text(5, 5, f'Both\n{n_both}', ha='center', va='center', fontsize=12, fontweight='bold')
ax1.text(7.5, 5, f'Only\nPsych\n{n_only_psych}', ha='center', va='center', fontsize=12, fontweight='bold')
ax1.text(5, 1.5, f'Neither: {n_neither}', ha='center', va='center', fontsize=11)

# Circle labels
ax1.text(2, 8.2, 'Seniors', ha='center', fontsize=12, fontweight='bold', color='blue')
ax1.text(8, 8.2, 'Psych Majors', ha='center', fontsize=12, fontweight='bold', color='red')

plt.show()


p_senior_or_psych_wrong = (n_senior + n_psych - n_both) / n_students
p_senior_or_psych_correct = (n_senior + n_psych) / n_students

print(f"\n💡 Key Insight:")
print(f"The overlap of {n_both} students must be subtracted to avoid double-counting!")
print(f"Without correction: {p_senior_or_psych_wrong:.0%} (wrong)")
print(f"With correction: {p_senior_or_psych_correct:.0%} (correct)")
```

### Part 4 - Multiplication Rule

**Independent events**: The occurrence of one event has no effect on the probability that the other event will occur.

**Multiplication Rule**: For independent events A and B:

$$P(A \text{ and } B) = P(A) \times P(B)$$

**Examples of Independent Events:**
- Two coin tosses (first toss doesn't affect second)
- Drawing cards with replacement
- Birth months of unrelated people
- Winning lottery numbers in different weeks

**Note**: The probability of two events occurring together is always smaller than or equal to the probability of either event occurring alone.

**Multiplication Rule Example**

A family has 2 children:

- Calculate the probability of having two boys.

```python
# Example: Probability of family composition
p_boy = 0.5
p_girl = 0.5

# Calculate probabilities for two children
p_two_boys = p_boy * p_boy
p_two_girls = p_girl * p_girl
p_one_of_each = 1 - p_two_boys - p_two_girls  # Or calculate directly

print("Family Composition Probabilities (2 children):")
print("="*50)
print(f"P(Boy) = {p_boy}")
print(f"P(Girl) = {p_girl}")
print(f"\nUsing Multiplication Rule:")
print(f"P(Two Boys) = P(B) × P(B) = {p_boy} × {p_boy} = {p_two_boys:.3f}")
print(f"P(Two Girls) = P(G) × P(G) = {p_girl} × {p_girl} = {p_two_girls:.3f}")
print(f"P(One of Each) = {p_one_of_each:.3f}")
```

### Part 5 - Conditional Probability

**Dependent events**

When the occurrence of one event affects the probability of the other event.

**Conditional probability** $P(B|A)$ is the probability of event B, given that event A has occurred.

For dependent events:
$$P(A \text{ and } B) = P(A) \times P(B|A)$$

**Examples:**
- Heights of siblings (if one is tall, the other is more likely to be tall)
- Card drawing without replacement
- Disease and test results
- Being a senior and being a psychology major (at a university)

**Key Insight**: When events are dependent, you must use the conditional probability of the second event, not its unconditional probability.

Consequently:

$$P(B|A) = \frac{P(A \text{ and } B)}{P(A)}$$

**Medical Testing Example**

Let's work through a realistic medical testing scenario using a **frequency analysis** approach, which makes conditional probabilities easier to understand.

```python
# Medical testing example with conditional probabilities
# Start with a population of 1,000 people
n_people = 1000
p_disease = 0.01  # 1% have the disease
p_test_pos_if_disease = 0.80  # True positive rate
p_test_pos_if_healthy = 0.10  # False positive rate

# Frequency analysis
n_disease = int(n_people * p_disease)
n_healthy = n_people - n_disease

# Among diseased people
n_test_pos_disease = int(n_disease * p_test_pos_if_disease)
n_test_neg_disease = n_disease - n_test_pos_disease

# Among healthy people
n_test_pos_healthy = int(n_healthy * p_test_pos_if_healthy)
n_test_neg_healthy = n_healthy - n_test_pos_healthy

# Total positive tests
n_test_pos_total = n_test_pos_disease + n_test_pos_healthy

print("Medical Testing Scenario (Frequency Analysis)")
print("="*60)
print(f"Total Population: {n_people}")
print(f"\nDisease Prevalence: {p_disease*100:.0f}%")
print(f"  People with disease: {n_disease}")
print(f"  Healthy people: {n_healthy}")
print(f"\nTest Characteristics:")
print(f"  True positive rate: {p_test_pos_if_disease*100:.0f}%")
print(f"  False positive rate: {p_test_pos_if_healthy*100:.0f}%")
print(f"\nTest Results:")
print(f"  Positive tests from diseased: {n_test_pos_disease}")
print(f"  Positive tests from healthy: {n_test_pos_healthy}")
print(f"  Total positive tests: {n_test_pos_total}")
```

**Now, calculate:**
- The probability of having disease, given that the patient tests positive for it.
- The probability of not having disease, given that the patient tests negative for it.

```python
# Calculate important conditional probabilities
prob_disease_given_pos = n_test_pos_disease / n_test_pos_total
prob_healthy_given_neg = n_test_neg_healthy / (n_test_neg_disease + n_test_neg_healthy)

print("\n" + "="*60)
print("KEY CONDITIONAL PROBABILITIES:")
print("="*60)
print(f"\nP(Disease | Positive Test) = {n_test_pos_disease}/{n_test_pos_total} = {prob_disease_given_pos:.3f}")
print(f"  → Only {prob_disease_given_pos*100:.1f}% of positive tests indicate disease!")
print(f"\nP(Healthy | Negative Test) = {n_test_neg_healthy}/{n_test_neg_disease + n_test_neg_healthy} = {prob_healthy_given_neg:.3f}")
print(f"  → {prob_healthy_given_neg*100:.1f}% of negative tests are truly healthy")

print(f"\n💡 Surprising Insight:")
print(f"Even with an 80% accurate test, most positive results ({(1-prob_disease_given_pos)*100:.1f}%)")
print(f"are FALSE POSITIVES because the disease is rare!")
```
