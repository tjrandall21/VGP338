### Part 8: Graph Objects: Layout Customization

```python
import numpy as np
from plotly.colors import qualitative

fig = go.Figure()

# Define color palette without px
continents = df['Continent'].unique()
palette = qualitative.Set3  # or qualitative.Plotly / qualitative.D3 etc.
colors = [palette[i % len(palette)] for i in range(len(continents))]

# Add scatter plot by continent
for i, continent in enumerate(continents):
    continent_data = df[df['Continent'] == continent]

    fig.add_trace(go.Scatter(x=continent_data['GDP_per_Capita'],
                             y=continent_data['Happiness_Score'],
                             mode='markers',
                             name=continent,
                             marker=dict(color=colors[i],
                                         size=12,
                                         line=dict(width=1, color='black')),
                             text=continent_data['Country'],
                             hovertemplate='<b>%{text}</b><br>' +
                             'GDP: %{x:.3f}<br>' +
                             'Happiness: %{y:.2f}' +
                             '<extra></extra>'))

# Add trend line
z = np.polyfit(df['GDP_per_Capita'], df['Happiness_Score'], 1)
p = np.poly1d(z)
x_trend = np.linspace(df['GDP_per_Capita'].min(),
                      df['GDP_per_Capita'].max(), 100)

fig.add_trace(go.Scatter(x=x_trend,
                         y=p(x_trend),
                         mode='lines',
                         name='Trend Line',
                         line=dict(color='black', width=2, dash='dash')))

# Add annotations for interesting points
top_country = df.loc[df['Happiness_Score'].idxmax()]
richest_country = df.loc[df['GDP_per_Capita'].idxmax()]

annotations = [
    dict(
        x=top_country['GDP_per_Capita'],
        y=top_country['Happiness_Score'],
        text=f"Happiest: {top_country['Country']}",
        showarrow=True,
        arrowhead=2,
        arrowcolor='green',
        bgcolor='lightgreen',
        bordercolor='green'
    ),
    dict(
        x=richest_country['GDP_per_Capita'],
        y=richest_country['Happiness_Score'],
        text=f"Richest: {richest_country['Country']}",
        showarrow=True,
        arrowhead=2,
        arrowcolor='gold',
        bgcolor='lightyellow',
        bordercolor='orange'
    )
]

# Update layout
fig.update_layout(
    title={
        'text': 'Happiness Analysis with Annotations',
        'x': 0.5,
        'font': {'size': 20, 'family': 'Arial Black'}
    },
    xaxis={
        'title': {'text': 'GDP per Capita', 'font': {'size': 14}},
        'gridcolor': 'lightgray',
        'showgrid': True
    },
    yaxis={
        'title': {'text': 'Happiness Score', 'font': {'size': 14}},
        'gridcolor': 'lightgray',
        'showgrid': True
    },
    annotations=annotations,
    width=700,
    height=500,
    legend=dict(
        bgcolor='rgba(255,255,255,0.8)',
        bordercolor='black',
        borderwidth=1
    )
)

fig.show()
```

### Part 9: Graph Objects: Interactive Choropleth Maps

Graph Objects excel at creating sophisticated visualizations with multiple layers of interactivity that would be difficult to achieve with Plotly Express.

This choropleth map demonstrates several advanced Graph Objects capabilities:

**Multi-Level Interactivity:**
- **Dynamic metric switching** via dropdown menus that update colors, scales, and hover information
- **Geographic filtering** to focus on specific continents
- **Rich hover tooltips** with multi-dimensional data display
- **Coordinated updates** where all elements change together seamlessly

**Technical Implementation:**
- **UpdateMenus**: Create dropdown controls that modify trace properties in real-time
- **Custom hover templates** with formatted data and conditional content
- **Geographic projections** with detailed coastline and ocean styling
- **Dynamic colorbar updates** that adapt title and scale to selected metrics

**Why Graph Objects?**
This level of coordinated interactivity requires direct control over:
- Trace property updates through `updatemenus`
- Custom geographic styling via `geo` layout properties
- Complex hover template formatting with `customdata`
- Precise positioning of UI controls and annotations

**Key Learning:**
Interactive choropleth maps showcase how Graph Objects provide the foundation for building dashboard-like visualizations where multiple UI elements work together to explore multidimensional geographic data.

```python
# Highly Interactive Choropleth Map (FIXED HOVER)
import plotly.graph_objects as go
import plotly.express as px
import pandas as pd
import numpy as np

# Prepare the data with all metrics
metrics_info = {
    'Happiness_Score': {'title': 'Happiness Score', 'colorscale': 'Viridis', 'format': '.2f'},
    'GDP_per_Capita': {'title': 'GDP per Capita', 'colorscale': 'Blues', 'format': '.3f'},
    'Social_Support': {'title': 'Social Support', 'colorscale': 'Greens', 'format': '.3f'},
    'Healthy_Life_Expectancy': {'title': 'Healthy Life Expectancy', 'colorscale': 'Oranges', 'format': '.3f'},
    'Freedom': {'title': 'Freedom Score', 'colorscale': 'Purples', 'format': '.3f'},
    'Generosity': {'title': 'Generosity Score', 'colorscale': 'Reds', 'format': '.3f'},
    'Trust_in_Government': {'title': 'Trust in Government', 'colorscale': 'YlOrRd', 'format': '.3f'}
}

# Create the main choropleth trace
fig = go.Figure()

# Add initial choropleth (FIXED hover)
fig.add_trace(
    go.Choropleth(
        locations=df['Country'],
        z=df['Happiness_Score'],
        locationmode='country names',
        colorscale='Viridis',
        marker_line_color='white',
        marker_line_width=0.8,
        text=df['Country'],  # ADD THIS: explicit text for hover
        colorbar=dict(title="Happiness Score",
                      thickness=20,
                      len=0.7,
                      x=1.02,
                      tickfont=dict(size=12)),
        hovertemplate='<b>%{text}</b><br>' +
                     'Happiness Score: %{z:.2f}<br>' +
                     'Continent: %{customdata[0]}<br>' +
                     'GDP per Capita: %{customdata[1]:.3f}<br>' +
                     'Social Support: %{customdata[2]:.3f}<br>' +
                     'Freedom: %{customdata[3]:.3f}<br>' +
                     '<extra></extra>',
        customdata=np.column_stack((df['Continent'],
                                    df['GDP_per_Capita'],
                                    df['Social_Support'],
                                    df['Freedom'])),
        name='World Happiness Data'
    )
)


# Create dropdown buttons for different metrics
dropdown_buttons = []
for metric, info in metrics_info.items():
    dropdown_buttons.append(
        dict(
            args=[
                {
                    'z': [df[metric]],
                    'colorscale': info['colorscale'],
                    'colorbar.title': info['title'],
                    'text': [df['Country']],  # ENSURE country names are always available
                    'hovertemplate': f'<b>%{{text}}</b><br>' +  # FIXED: use %{{text}}
                                   f'{info["title"]}: %{{z:{info["format"]}}}<br>' +
                                   'Continent: %{customdata[0]}<br>' +
                                   'GDP per Capita: %{customdata[1]:.3f}<br>' +
                                   'Social Support: %{customdata[2]:.3f}<br>' +
                                   'Freedom: %{customdata[3]:.3f}<br>' +
                                   '<extra></extra>'
                }
            ],
            label=info['title'],
            method='restyle'
        )
    )

# Create continent filter buttons (FIXED to include text)
continent_buttons = []
continents = ['All'] + list(df['Continent'].unique())

for continent in continents:
    if continent == 'All':
        # Show all countries
        continent_buttons.append(
            dict(
                args=[{
                    'locations': [df['Country'].tolist()],
                    'z': [df['Happiness_Score'].tolist()],
                    'text': [df['Country'].tolist()],  # FIXED: include country names
                    'customdata': [np.column_stack((
                        df['Continent'],
                        df['GDP_per_Capita'],
                        df['Social_Support'],
                        df['Freedom']
                    ))]
                }],
                label=continent,
                method='restyle'
            )
        )
    else:
        # Filter by continent
        continent_df = df[df['Continent'] == continent]
        continent_buttons.append(
            dict(
                args=[{
                    'locations': [continent_df['Country'].tolist()],
                    'z': [continent_df['Happiness_Score'].tolist()],
                    'text': [continent_df['Country'].tolist()],  # FIXED: include country names
                    'customdata': [np.column_stack((
                        continent_df['Continent'],
                        continent_df['GDP_per_Capita'],
                        continent_df['Social_Support'],
                        continent_df['Freedom']
                    ))]
                }],
                label=continent,
                method='restyle'
            )
        )

# Update layout with interactive controls
fig.update_layout(
    title={
        'text': '<b>🌍 Interactive World Happiness Atlas</b><br>' +
               '<sub>Select metrics and continents using the dropdown menus</sub>',
        'x': 0.5,
        'xanchor': 'center',
        'font': {'size': 20}
    },

    # Dropdown for metrics
    updatemenus=[
        {
            'buttons': dropdown_buttons,
            'direction': 'down',
            'showactive': True,
            'x': 0.02,
            'y': 0.98,
            'xanchor': 'left',
            'yanchor': 'top',
            'bgcolor': 'rgba(255,255,255,0.9)',
            'bordercolor': 'black',
            'borderwidth': 1,
            'font': {'size': 12}
        },
        {
            'buttons': continent_buttons,
            'direction': 'down',
            'showactive': True,
            'x': 0.02,
            'y': 0.85,
            'xanchor': 'left',
            'yanchor': 'top',
            'bgcolor': 'rgba(255,255,255,0.9)',
            'bordercolor': 'black',
            'borderwidth': 1,
            'font': {'size': 12}
        }
    ],

    # Geographic settings
    geo=dict(
        showframe=False,
        showcoastlines=True,
        coastlinecolor="RebeccaPurple",
        coastlinewidth=2,
        projection_type='natural earth',
        showlakes=True,
        lakecolor='lightblue',
        showocean=True,
        oceancolor='lightblue',
        showland=True,
        landcolor='rgb(243, 243, 243)',
        showcountries=True,
        countrycolor='white',
        countrywidth=0.5
    ),

    width=1200,
    height=700,

    # Add annotations for controls
    annotations=[
        dict(
            text="📊 Select Metric:",
            x=0.02, y=1.02,
            xref="paper", yref="paper",
            showarrow=False,
            font=dict(size=12, color="darkblue"),
            bgcolor="rgba(255,255,255,0.8)"
        ),
        dict(
            text="🌎 Filter by Continent:",
            x=0.02, y=0.89,
            xref="paper", yref="paper",
            showarrow=False,
            font=dict(size=12, color="darkblue"),
            bgcolor="rgba(255,255,255,0.8)"
        ),
        dict(
            text="🔍 Hover over countries for details<br>📈 Switch metrics with dropdown<br>🗺️ Pan and zoom to explore",
            x=0.98, y=0.02,
            xref="paper", yref="paper",
            showarrow=False,
            font=dict(size=10),
            bgcolor="rgba(255,255,255,0.9)",
            bordercolor="gray",
            borderwidth=1,
            xanchor="right",
            yanchor="bottom"
        )
    ]
)
fig.show()
```
