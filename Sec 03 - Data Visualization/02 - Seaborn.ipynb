```python
# Load a tiny sample dataset
tips = sns.load_dataset("tips")

# A very simple scatter plot
sns.scatterplot(data=tips,
                x="total_bill",
                y="tip")
```

### Part 2 - Seaborn Categories

<a href="https://imgbb.com/"><img src="https://i.ibb.co/Ng2TbrmG/overview.png" alt="overview" border="0"></a>

```python
# Categorical data: boxplot
ax = sns.boxplot(data=tips,
                 x="day",
                 y="total_bill")


ax.set_title("Categorical: Boxplot of total bill by day")
```

```python
# Distributional: histplot
ax = sns.histplot(data=tips,
                  x="day",
                  bins=4,
                  kde=True)

ax.set_title("Distributional: Histogram of total bill")
```

```python
# Regression: regplot
ax = sns.regplot(data=tips,
                 x="total_bill",
                 y="tip")


ax.set_title("Regression: Regression line of total bill vs tip")
```

```python
# Temp
# Matrix: heatmap of correlations
corr = tips.corr(numeric_only=True)  # restrict to numeric columns
ax = sns.heatmap(corr,
                 annot=True,
                 cmap="coolwarm")
ax.set_title("Correlation heatmap")
```

### Part 3 - MultiView Functions

```python
# 1) jointplot: bivariate view + marginals (scatter + hist/kde)
sns.jointplot(data=tips,
              x="total_bill",
              y="tip",
              hue="sex",
              kind="scatter",  # try "kde" for density view
              height=4)
```

```python
# 2) pairplot: pairwise across a few numeric variables

cols = ["total_bill", "tip", "size"]
sns.pairplot(data=tips[cols + ["sex"]],  # include a hue column for classes
             hue="sex",
             corner=True,
             height=1.8)
```

### Part 4 - Relational Plots

```python
# Customization
sns.set_theme(style="whitegrid",
              palette="deep",
              context="notebook")

ax = sns.scatterplot(data=tips,
                     x="total_bill",
                     y="tip",
                     hue="time",
                     size="size",
                     style="smoker")

ax.set_title("Multidimensional Scatter Plot of Dining Experiences")
```

```python
# Line Plots
fmri = sns.load_dataset("fmri")

ax = sns.lineplot(data=fmri,
                  x="timepoint",
                  y="signal")

ax.set_title("fMRI signal over time")
ax.set_xlabel("timepoint")
ax.set_ylabel("signal")
```

```python
# Advanced line: groups by event and region + markers + error bars
ax = sns.lineplot(data=fmri,
                  x="timepoint",
                  y="signal",
                  hue="event",            # color encodes event
                  style="region",         # line style/markers encode region
                  markers=True,           # draw markers at data points
                  dashes=False,           # solid lines for readability
                  errorbar="se",          # show standard error band ('ci', 'pi', 'se', 'sd')
                  estimator="mean")       # aggregate central tendency (default is mean)

ax.set_title("Advanced line: event × region with markers and SE band")
ax.set_xlabel("timepoint")
ax.set_ylabel("signal")
```

### Part 5 - Figure Levels

```python
import seaborn as sns

tips = sns.load_dataset("tips")

# Advanced relplot: facets + hue + size + layout tuning
g = sns.relplot(data=tips,
                x="total_bill",
                y="tip",
                hue="sex",               # color encodes sex
                size="size",
                sizes=(20, 120),  # point size encodes party size
                col="time",
                col_wrap=2,   # facets across days (wrapped to 2 columns)
                kind="scatter",
                palette="mako",
                alpha=0.7,
                height=3.0,
                aspect=1.0,)

# Titles and labels
g.set_titles(col_template="{col_name}")
g.set_axis_labels("Total bill ($)", "Tip ($)")
```

### Part 6 - Distribution Plots

```python
sns.set_theme(style='whitegrid')

penguins = sns.load_dataset('penguins')

g = sns.displot(data=penguins,
                x='flipper_length_mm',
                col='species',
                kind='hist',
                bins=20,
                height=3,
                aspect=1.0)

g.set_axis_labels('Flipper length (mm)', 'Count')
```

```python
# Customization
ax = sns.histplot(data=penguins, # sns.kdeplot
                  x='flipper_length_mm',
                  bins=20,
                  hue='species',
                  element='step',
                  stat='density',
                  common_norm=False,
                  kde=True)

ax.set_title('Flipper Length by Species (density, 30 bins)')
ax.set_xlabel('Flipper length (mm)')
ax.set_ylabel('Density')
```

```python
# Step Function : Empirical Cumulative Distribution Function
ax = sns.ecdfplot(data=penguins,
                  x='flipper_length_mm',
                  hue='species',
                  complementary=True)

ax.set_title('Complementary CDF by Species')
ax.set_xlabel('Flipper length (mm)')
ax.set_ylabel('1 - Cumulative proportion')
```

### Part7 - Categorical Plots

```python
penguins = sns.load_dataset("penguins") # Continuous vs. Categorical

ax = sns.stripplot(data=penguins,
                   x="species",
                   y="body_mass_g",
                   hue="sex",
                   dodge=True)

ax.set_title("Body Mass Distribution of Penguin Species by Sex")
ax.set_xlabel("Penguin Species")
ax.set_ylabel("Body Mass (g)")
```

```python
# Swarm Plots
ax = sns.swarmplot(data=penguins,
                   x="species",
                   y="flipper_length_mm",
                   hue='sex',
                   size=4)

ax.set_title("Flipper Length Distribution of Penguin Species")
ax.set_xlabel("Penguin Species")
ax.set_ylabel("Flipper Length (mm)")
```

### Part 12 - Heat Maps

```python
import matplotlib.pyplot as plt
```

```python
# Pivot a simple table: monthly passengers by year
flights = sns.load_dataset("flights")
flights = flights.pivot(index="month",
                        columns="year",
                        values="passengers")

plt.figure(figsize=(8, 6))   # width=8, height=6 in inches
ax = sns.heatmap(flights,
                 annot=True,               # write values in cells
                 fmt="d",                 # integer formatting
                 linewidths=0.5,           # grid lines
                 linecolor="white",
                 cbar_kws={"label": "Passengers"},  # colorbar label
                 cmap="coolwarm")

# Rotate tick labels instead of ticks
ax.set_xticklabels(ax.get_xticklabels(), rotation=45, ha="center")
ax.set_yticklabels(ax.get_yticklabels(), rotation=0)
ax.set_title("Flights heatmap with annotations and custom palette")
```

### Part 13 - Facet Grid

```python
import seaborn as sns

# Load the tips dataset
tips = sns.load_dataset("tips")

# Create a FacetGrid
g = sns.FacetGrid(tips,
                  col="time", # dinner / lunch
                  row="sex", # male / female
                  height=2,
                  aspect=1.5)

# Map a scatterplot onto the grid
g.map(sns.scatterplot, "total_bill", "tip")
```

```python
g = sns.FacetGrid(tips,
                  col="day",
                  height=2,
                  aspect=1.5,
                  col_wrap=2)

g.map(sns.histplot, "total_bill",
      kde=True,
      color="skyblue",
      line_kws=dict(color="navy"))

g.set_axis_labels("Total Bill ($)", "Count")
g.set_titles("{col_name}")
```
