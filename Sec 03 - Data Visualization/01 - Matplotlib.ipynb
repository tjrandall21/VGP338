```python
# put 50 properties together
# with areas of 500 to 1000 sqft
# with prices of 400_000 to 1_000_000
# with 0 to 5 bedrooms
# with 1st to 30th floor
# Represent the whole thing within a scatter plot

n = 50
ar = np.random.randint (500, 1000, n)
pr = np.random.randint (400_000, 1_000_000, n)
bd = np.random.randint (0, 4, n) + 1 #
fl = np.random.randint (1, 30, n)

fig, ax = plt.subplots()
sc = ax.scatter(ar, pr, c=fl, s=bd * 20, cmap='turbo', alpha=0.7)

cbar = fig.colorbar(sc)
cbar.set_label('Floor 0-30')

# Add a simple least-squares line
z = np.polyfit(ar, pr, 1)
ax.plot(ar, np.poly1d(z)(ar), 'r--', label='Trend')

ax.set_title('Scatter plot of prices, areas, bedrooms and floors')
ax.set_xlabel('Area (sqft)')
ax.set_ylabel('Price (CAD)')
ax.legend()
plt.show()
```

```python
# Grouped Bars
categories = ['20s', '30s', '40s', '50s', '60s'] # strings
men = [60, 62, 77, 82, 70]
women = [56, 58, 57, 60, 62]
dogs = [40,46,63,52,80]

fig, ax = plt.subplots()

# Create grouped bar chart
x = np.arange(len(categories))
width = 0.35

rects1 = ax.bar(x - width/3, men, width/3, label='Men', color='skyblue')
rects2 = ax.bar(x, women, width/3, label='Women', color='lightgreen')
rects3 = ax.bar(x + width/3, dogs, width/3, label='Dogs', color='orange')


# Customize the plot
ax.set_title("Customized Grouped Bar Chart")
ax.set_xlabel("Age")
ax.set_ylabel("Weight")
ax.set_xticks(x)
ax.set_xticklabels(categories)
ax.legend()

# Add a grid
ax.grid(True, axis='y', linestyle='--', alpha=0.7)

# Add value labels
def autolabel(rects):
    for rect in rects:
        height = rect.get_height()
        ax.annotate(f'{height}',
                    xy=(rect.get_x() + rect.get_width() / 2, height),
                    xytext=(0, 3),  # 3 points vertical offset
                    textcoords="offset points",
                    ha='center', va='bottom')

autolabel(rects1)
autolabel(rects2)
autolabel(rects3)
```

### Part 9 - Histogram

```python
# Create multiple subplots to show different distributions
fig, axes = plt.subplots(1, 2, figsize=(8, 4))
fig.suptitle('Histograms of Different Random Distributions', fontsize=16, fontweight='bold')

# Normal distribution
data_normal = rng.normal(loc=0, scale=0.3, size=1_000)

axes[0].hist(data_normal, bins=7, color='#55A868', edgecolor='white')

axes[0].set_title('Normal Distribution (μ=0, σ=0.3)')
axes[0].set_xlabel('Value')
axes[0].set_ylabel('Frequency')
axes[0].set_xlim(-1, 1)

# Uniform distribution
data_uniform = rng.uniform(low=-1, high=1, size=1_000)
axes[1].hist(data_uniform, bins=7, color='#4ECDC4', edgecolor='white')
axes[1].set_title('Uniform Distribution (-1 to 1)')
axes[1].set_xlabel('Value')
axes[1].set_ylabel('Frequency')
axes[1].set_xlim(-1, 1)


fig.tight_layout()
```

```python
# Multiple Datasets
men = np.random.normal(175, 3, 1000) # Height of men
women = np.random.normal(165, 2, 1000) # Height of women

fig, ax = plt.subplots()

# Create overlapping histograms
ax.hist(men, bins=10, alpha=0.7, label='Men Height', edgecolor='black')
ax.hist(women, bins=10, alpha=0.7, label='Women Height', edgecolor='black')

# Customize the plot
ax.set_title("Histogram comparing height of men and women")
ax.set_xlabel("Heights")
ax.set_ylabel("Frequency")
ax.legend()

# Add a grid
ax.grid(True, linestyle='--', alpha=0.7)

# Add vertical lines for means
ax.axvline(men.mean(), color='r', linestyle='dashed', linewidth=2, label='Mean Men Height')
ax.axvline(women.mean(), color='g', linestyle='dashed', linewidth=2, label='Mean Women Height')

# Update legend
ax.legend()
```

```python
# Example of a KDE plot
from scipy import stats

fig, ax = plt.subplots()

# Plot histogram and KDE (Kernel Density Estimation)
ax.hist(women, bins=30, density=True, alpha=0.7, edgecolor='black')
kde = stats.gaussian_kde(women)
x_range = np.linspace(women.min(), women.max(), 100)
ax.plot(x_range, kde(x_range), label='KDE')

ax.set_title("Histogram with KDE")
ax.set_xlabel("Values")
ax.set_ylabel("Density")
ax.legend()
```

### Part 10 - Combining Plots

```python
# Generate temperature data for a year
days = np.arange(1, 366)  # Day of year (1-365)

# Simulate temperature with seasonal variation + random noise
temp_base = 20 + 15 * np.sin(2 * np.pi * (days - 80) / 365)  # Seasonal cycle
temperature = temp_base + rng.normal(0, 3, len(days))  # Add daily variation

# Create subplots to show different views of the same data
fig, axes = plt.subplots(1, 2, figsize=(8, 4))

# Left plot: Temperature over time (trend)
axes[0].plot(days, temperature, color='#FF6B6B', linewidth=1, alpha=0.8)
axes[0].set_title('Daily Temperature Throughout the Year')
axes[0].set_xlabel('Day of Year')
axes[0].set_ylabel('Temperature (°C)')
axes[0].grid(True, alpha=0.3)
axes[0].set_xlim(1, 365)

# Right plot: Temperature distribution (frequency)
axes[1].hist(temperature, bins=40, color='#4ECDC4', edgecolor='white', alpha=0.7)
axes[1].set_title('Temperature Distribution')
axes[1].set_xlabel('Temperature (°C)')
axes[1].set_ylabel('Number of Days')
axes[1].axvline(temperature.mean(),
                color='red',
                linestyle='--',
                label=f'Mean: {temperature.mean():.1f}°C')
axes[1].legend()

fig.suptitle('Annual Temperature Analysis: Time Series vs Distribution')
fig.tight_layout()
plt.show()
```

```python
import requests
from io import BytesIO

image_url = "https://www.spjain.org/hubfs/Imported_Blog_Media/1-steve-jobs.jpg"

response = requests.get(image_url)
response.raise_for_status()  # ensure download succeeded

pil_image = Image.open(BytesIO(response.content))
image = np.array(pil_image)
```

```python
# Display the images using imshow()
fig, axes = plt.subplots(1, 2, figsize=(8, 4))

# Original color image
axes[0].imshow(image)
axes[0].set_title('Steve Jobs (3D Array)\nShape: ' + str(image.shape))
axes[0].axis('off')

# Grayscale version
grayscale_image = np.mean(image, axis=2).astype(np.uint8)
axes[1].imshow(grayscale_image, cmap='gray')
axes[1].set_title('Grayscale Version (2D Array)\nShape: ' + str(grayscale_image.shape))
axes[1].axis('off')

plt.tight_layout()
plt.show()

```

```python
x = np.linspace(-2*np.pi, 2*np.pi, 100)
y = np.linspace(-2*np.pi, 2*np.pi, 100)
X, Y = np.meshgrid(x, y)
Z = np.sin(X) * np.cos(Y)

# Compare different colormaps
colormaps = ['viridis', 'plasma', 'coolwarm', 'seismic'] # plt.style.available

fig, axes = plt.subplots(2, 2, figsize=(12, 10))
axes = axes.flatten()

for i, cmap in enumerate(colormaps):
    im = axes[i].imshow(Z, cmap=cmap)
    axes[i].set_title(f'Colormap: {cmap}')
    fig.colorbar(im, ax=axes[i])

fig.tight_layout()
```

```python
# Value range control with vmin and vmax
fig, axes = plt.subplots(1, 4, figsize=(16, 4))

# Original full range
im1 = axes[0].imshow(grayscale_image, cmap='gray')
axes[0].set_title('Default range\n(vmin=auto, vmax=auto)')
axes[0].axis('off')
fig.colorbar(im1, ax=axes[0])

# Clipped high values
im2 = axes[1].imshow(grayscale_image, cmap='gray', vmin=0, vmax=150)
axes[1].set_title('vmin=0, vmax=150\n(bright areas clipped)')
axes[1].axis('off')
fig.colorbar(im2, ax=axes[1])

# Clipped low values
im3 = axes[2].imshow(grayscale_image, cmap='gray', vmin=100, vmax=255)
axes[2].set_title('vmin=100, vmax=255\n(dark areas clipped)')
axes[2].axis('off')
fig.colorbar(im3, ax=axes[2])

# Narrow range for contrast enhancement
im4 = axes[3].imshow(grayscale_image, cmap='gray', vmin=50, vmax=200)
axes[3].set_title('vmin=50, vmax=200\n(enhanced contrast)')
axes[3].axis('off')
fig.colorbar(im4, ax=axes[3])

fig.suptitle('Value Range Control (vmin, vmax)', fontsize=14, y=1.02)
fig.tight_layout()
```

### Part 13 - Contours

```python
# Create sample data for contour plotting
x = np.linspace(-5, 5, 50)
y = np.linspace(-5, 5, 50)
X, Y = np.meshgrid(x, y)
Z = np.sin(np.sqrt(X**2 + Y**2))

# Display both image and contour plots side by side
fig, axes = plt.subplots(1, 2, figsize=(8, 4))

# Left: Display as image
im = axes[0].imshow(Z) # extent=[-5, 5, -5, 5], origin='lower', cmap='viridis'
axes[0].set_title('Same Data as Image')
axes[0].set_xlabel('X')
axes[0].set_ylabel('Y')
fig.colorbar(im, ax=axes[0])

# Right: Display as contour plot
contours = axes[1].contour(X, Y, Z, levels=5)
axes[1].clabel(contours, inline=True, fontsize=8)

axes[1].set_title('Same Data as Contour Lines')
axes[1].set_xlabel('X')
axes[1].set_ylabel('Y')
axes[1].grid(True, alpha=0.3)

fig.suptitle('Comparing Image and Contour Representations', fontsize=14)
fig.tight_layout()
plt.show()
```

```python
# Filled contour plot
plt.figure(figsize=(10, 8))
contourf = plt.contourf(X, Y, Z, levels=5, cmap='RdYlBu_r')

# plt.imshow(Z)

plt.colorbar(contourf, label='Z values')
contours = plt.contour(X, Y, Z, levels=10, colors='black', alpha=0.4, linewidths=0.5)
plt.title('Filled Contour Plot with Overlay Lines')
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```

```python
# Create more complex data
x = np.linspace(0, 4*np.pi, 100)
y = np.linspace(0, 4*np.pi, 100)
X, Y = np.meshgrid(x, y)
Z = np.sin(X) * np.sin(Y) * np.exp(-((X-2*np.pi)**2 + (Y-2*np.pi)**2)/8)

plt.figure(figsize=(10, 8))

# Image as background - capture the image object
im = plt.imshow(Z, extent=[0, 4*np.pi, 0, 4*np.pi], origin='lower',
                cmap='viridis', alpha=0.8)

# Contour lines overlay
contours = plt.contour(X, Y, Z, levels=8, colors='white', alpha=0.7, linewidths=1.5)
plt.clabel(contours, inline=True, fontsize=10, colors='white')

# Associate colorbar with the image object
plt.colorbar(im, label='Intensity')
plt.title('Combined Image and Contour Visualization')
plt.xlabel('X')
plt.ylabel('Y')
```

### Part 14 - Co-relation Matrix

```python
# Create sample correlation matrix
variables = ['Var A', 'Var B', 'Var C', 'Var D', 'Var E']
n_vars = len(variables)

# Generate random correlation matrix
correlation_matrix = np.random.rand(n_vars, n_vars)

plt.figure(figsize=(8, 6))
im = plt.imshow(correlation_matrix)


# Add text annotations
for i in range(n_vars):
    for j in range(n_vars):
        text = plt.text(j, i, f'{correlation_matrix[i, j]:.2f}',
                       ha="center", va="center", color="black", fontweight='bold')

plt.colorbar(im, label='Correlation Coefficient')
plt.xticks(range(n_vars), variables)
plt.yticks(range(n_vars), variables)
plt.title('Correlation Matrix Heatmap')
plt.tight_layout()
plt.show()
```
