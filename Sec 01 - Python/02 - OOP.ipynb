### Part 3 - Python modules

```python
pi = 22 / 7  # Defining the constant for pi

def sphereArea(r):
    '''Calculates the surface area of a sphere with radius r

        Parameters
        ----------
        r : float
            The radius of the sphere

        pi (constant): a constant, defined as 22/7

        Returns
        -------
        area
            The surface area of the sphere
    '''
    area = 4 * pi * r**2
    return area

def sphereVolume(r):
    '''Calculates the volume of a sphere with radius r

        Parameters
        ----------
        r : float
            The radius of the sphere

        pi (constant): a constant, defined as 22/7

        Returns
        -------
        volume
            The volume of the sphere
    '''
    volume = (4 / 3) * (pi * r**3)
    return volume
```

### Part 5 - Return vs Yield

```python
def square_list(numbers):
    result = []
    for num in numbers:
        result.append(num * num)
    return result

squares = square_list([1, 2, 3])
print(squares) # Output: [1, 4, 9]
# The entire list of squares is created and stored in memory before being returned.
```

```python
# Produce a sequence of values one at a time (lazily),
# Significantly reduces memory usage

def square_generator(numbers):
    for num in numbers:
        yield num * num

squares_gen = square_generator([1, 2, 3])
print(squares_gen) # Output: <generator object square_generator at ...>

# Iterate over the generator to get values one by one
for square in squares_gen:
    print(square)

# The entire sequence is never in memory at the same time.
```
