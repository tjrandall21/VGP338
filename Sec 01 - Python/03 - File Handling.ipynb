### Part 2 - Reading CSV Files


```python
import csv
data_directory = 'data'
file_name = 'bank.csv'  
file_path = os.path.join(data_directory, file_name)

file_obj = open(file_path, 'r', encoding='utf-8')
csv_reader = csv.DictReader(file_obj, delimiter=',', quotechar='"')
print(csv_reader)

file_obj.close()  
```

```python
# enumerate csv file

file_obj=open(file_path, 'r', encoding='utf-8')
csv_reader = csv.DictReader(file_obj, delimiter=',', quotechar='"')
for idx, row in enumerate(csv_reader):
    print(idx, row)
    if idx == 4:
        break
```

```python
output_file_path = os.path.join("data", 'output_csv.csv')
input_file_path = os.path.join("data", 'bank.csv')

with open(output_file_path, mode='w', encoding='utf-8') as output_file:
    fieldnames = ['marital', 'balance']
    csv_writer = csv.DictWriter(output_file, fieldnames=fieldnames)
    csv_writer.writeheader()

    # Now we read the input file, compare the price, and write to the output
    with open(input_file_path, 'r', encoding='utf-8') as input_file:
        csv_reader = csv.DictReader(input_file)
        for idx, row in enumerate(csv_reader):
            price = float(row['balance'])
            if price < 1000000:
                address = row['job']
                balance = row['balance']
                csv_writer.writerow(
                    {'marital': address,
                     'balance': balance}
                )
print(f'Successfully written the output file at {output_file_path}')
```
