### Part 2 - "Big O" Notation

Big O notation does not measure the exact computation time of an algorithm; rather, it describes how the runtime or resource usage grows relative to the input size as the input size increases.

1. O(1) - Constant Time: The algorithm's runtime remains constant regardless of the input size. Returning the length of a list is an example of O(1) operation.

2. O(log n) - Logarithmic Time: The runtime increases logarithmically with the input size. This is often seen in algorithms that divide the problem size in half at each step (e.g., binary search).

3. O(n) - Linear Time: The runtime increases linearly with the input size. If you have to iterate through a list of n elements once, it will take O(n) time.

4. O(n log n) - Linearithmic Time: The runtime grows slightly faster than linear time. Efficient sorting algorithms like merge sort and heapsort have this complexity.

5. O(n<sup>2</sup>) - Quadratic Time: The runtime increases quadratically with the input size. Nested loops often result in O(n<sup>2</sup>) complexity. If you have to iterate through a list of n elements n times, it will be O(n<sup>2</sup>).

6. O(2<sup>n</sup>) - Exponential Time: The runtime doubles with each addition to the input size. These algorithms become impractical for even moderately sized inputs.

For example, if we suddenly had 10 times or even 1,000 times more input data to work with:


| Complexity     | Small Input (n=10) | Large Input (n=1000) |
|----------------|--------------------|----------------------|
| O(1)          | Constant           | Constant             |
| O(log n)      | ~3                 | ~10                  |
| O(n)          | 10                 | 1000                 |
| O(n log n)    | ~30                | ~10,000              |
| O(n²)         | 100               | 1,000,000            |
| O(2ⁿ)         | 1024              | ~10³⁰⁰             |

- When analyzing time complexity using Big O, we focus on the dominant term and ignore constant factors and lower-order terms. 
- For example, an algorithm with a runtime of O(2n<sup>2</sup> + 5n + 10) is considered O(n<sup>2</sup>) because as n grows very large, the n<sup>2</sup> term dominates the other terms and the constant factor 2 becomes insignificant in comparison to the rate of growth.

#### O(1) - *Constant Time*
Accessing an element by index in a list always takes constant time, regardless of the list size.

```python
lst = [10, 20, 30, 40, 50]
element = lst[2]  # Always takes O(1) time
length = len(lst) # Always takes O(1) time
```

#### O(n) - *Linear Time*
Iterating through a list to find the maximum element requires visiting every element.

```python
def linear_search(lst, target):
    for i, item in enumerate(lst):
        if item == target:
            return i  # Return the index if the target is found
    return None  # Return None if the target is not in the list

# Example usage
lst = [3, 1, 4, 1, 5]
target = 4
result = linear_search(lst, target)  # O(n)
print(f"Target found at index: {result}")
```

```python
lst = [3, 1, 4, 1, 5]
max_value = max(lst)  # Iterates through all elements, O(n)
```

#### O(n²) - *Quadratic Time*
Checking all pairs of elements in a list involves nested loops.

```python
lst = [1, 2, 3]
for i in lst:
    for j in lst:
        print(i, j)  # Nested iteration results in O(n²)
```

#### O(log n) - *Logarithmic Time*
Searching for an element in a sorted list using binary search.

```python
def binary_search(lst, target):
    left, right = 0, len(lst) - 1
    while left <= right:
        mid = (left + right) // 2
        if lst[mid] == target:
            return mid
        elif lst[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1

binary_search([1, 2, 3, 4, 5], 4)  # O(log n)
```

#### O(2ⁿ) - *Exponential Time*
brute-force subset generation.

```python
def fibonacci_recursive(n):
    if n <= 1:
        return n
    return fibonacci_recursive(n-1) + fibonacci_recursive(n-2)  # Two branches per call
```

### Part 3 - Space Complexity

Space complexity measures *how much memory an algorithm uses as the input size grows*. This includes both:
- **Input space**: Memory required to store the input data.
- **Auxiliary space**: extra space used by the algorithm.

However, *Auxiliary space is the primary focus of space complexity analysis* because input data size is usually fixed or predetermined, but excessive auxiliary space can lead to inefficiency or even program crashes due to memory exhaustion.

#### Practical exmaple

Organizing books in a library. We’ll examine two different "algorithms" for this task:

1. Table Sorting: Spreading out all books on a large table to sort them.

2. Cart Sorting: Using a small cart that holds only 2-3 books at a time to sort them by shuffling between shelves.

#### **Table Sorting**

*Space Usage*: The table must be large enough to hold every book simultaneously. If there are n books, the required space scales linearly with n. For example, a table 10 meters long might hold 1000 books, while a table 20 meters long would hold 2000 books.

- Pros: Efficient for small numbers of books (quick to organize when the table is available).
- Cons: Impractical for large libraries. A table big enough for 1 million books would be impossibly large!

##### **Cart Sorting**

*Space Usage*: The cart’s size is fixed. Whether the library has 100 or 100,000 books, the cart occupies the same physical space.

- Pros: Works for libraries of any size. No need for massive tables.
- Cons: More time-consuming (many trips between shelves), but space-efficient.

- This analogy highlights a crucial point: the efficiency of an "algorithm" can vary dramatically depending on the input size.
- For small libraries, table sorting is clearly the superior choice. However, for extremely large libraries, the cart method becomes necessary because the table method would require an impractically large amount of space.
- This demonstrates the importance of analyzing how an algorithm’s space usage scales with input size, rather than just considering its performance on a single, specific input.

#### Big O Notation for Space Complexity

Similar to time complexity, Big O notation for space complexity describes *how the memory usage of an algorithm grows as the input size (n) increases*. It helps us quantify the maximum additional space (auxiliary space) an algorithm requires to solve a problem.

For example, in the library book sorting problem, table sorting space complexity is O(n), while cart sorting method is O(1).

Here's an ordered breakdown of common Big O space complexities, from smallest to largest memory growth:

1. O(1) - *Constant Space*: The algorithm uses the same amount of additional memory regardless of the input size.

2. O(log n) - *Logarithmic Space*: The memory requirement grows logarithmically with the input size.

3. O(n) - *Linear Space*: The memory usage grows proportionally with the input size.
Example: Storing an array of size n requires O(n) space. Similarly, an algorithm that keeps all input elements in memory uses linear space.

4. O(n log n) - *Linearithmic Space*: The memory usage grows faster than linear but slower than quadratic.
Example: Certain divide-and-conquer algorithms, like merge sort, require O(nlogn) space for temporary storage during the merging process.

5. O(n²) - *Quadratic Space*: The memory usage grows quadratically with the input size.
Example: Storing a 2D table (e.g., a matrix for dynamic programming) often requires O(n²) space.

6. O(2ⁿ) - Exponential Space
The memory requirement doubles with each addition to the input size.
Example: Recursive algorithms for problems like the traveling salesman (without optimization) can use O(2ⁿ) space for their call stack and memoization.

##### O(1) - *Constant Space*
Sum of an array requires storing the sum itself, no matter how large the array is.

```python
def sum_array(arr):
    total = 0  # Single variable for accumulation
    for num in arr:
        total += num
    return total
```

##### O(n) - *Linear Space*
Storing an array of size n requires O(n) space e.g., storing all elements in an additional.

```python
def reverse_array(arr):
    # A new list of size n is created
    return arr[::-1]

arr = [1, 2, 3, 4, 5]
print(reverse_array(arr))  # [5, 4, 3, 2, 1]
```

##### O(n²) - *Quadratic Space*
Storing a 2D table (e.g., a matrix for dynamic programming) often requires O(n²) space.

```python
def multiplication_table(n):
    # Create a 2D list (n x n) to store the table
    table = [[0] * n for _ in range(n)]

    for i in range(n):
        for j in range(n):
            table[i][j] = (i + 1) * (j + 1)

    return table

# Example usage
n = 10
table = multiplication_table(n)

# Print the table with proper formatting
print("Multiplication Table:")
print("-" * (n * 5))
for row in table:
    print(" | ".join(f"{num:2}" for num in row))
print("-" * (n * 5))
```

##### O(2ⁿ) - Exponential Space
storing all subsets

```python
def generate_subsets(nums):
    subsets = [[]]
    for num in nums:
        new_subsets = []
        for subset in subsets:
            new_subsets.append(subset + [num])
        subsets += new_subsets
    return subsets

generate_subsets([1, 2, 3])
```

### Part 4 - Time-Space Complexity Trade-off

- In algorithm design, there’s often a tug-of-war between time and space efficiency. 
- Optimizing for one usually comes at the expense of the other. 
- This fundamental trade-off forces programmers to make strategic choices based on the problem’s constraints, hardware limitations, and application requirements. 

**Why the Trade-Off Exists?**

- Time Efficiency: Faster algorithms often precompute, cache, or store intermediate results to avoid redundant work. This requires extra space.

- Space Efficiency: Memory-constrained algorithms avoid storing extra data, but this often forces them to recompute values or use slower operations.

Approach 1: Brute-Force (Slow, Minimal Space)

```python
def has_duplicates_slow(arr):
    # Compare every pair of elements
    for i in range(len(arr)):
        for j in range(i + 1, len(arr)):
            if arr[i] == arr[j]:
                return True
    return False
```

- Time Complexity: For a list of size n, this checks n*(n-1)/2 pairs. **O(n²)**
- Space Complexity: No extra memory is used. **O(1)**

Approach 2: Sort First (Faster, More Space)

```python
def has_duplicates_fast(arr):
    # Create a sorted copy of the list (uses extra space)
    sorted_arr = sorted(arr)  # O(n) space
    # Check adjacent elements in the sorted list
    for i in range(len(sorted_arr) - 1):
        if sorted_arr[i] == sorted_arr[i + 1]:
            return True
    return False
```

### Part 4 - Case Study: Bubble Sort vs. Merge Sort 

- Sorting algorithms are fundamental to programming, but their efficiency varies dramatically.
- Let’s compare Bubble Sort (a simple "normal" sorting method) with Merge Sort (a more advanced algorithm) to highlight key differences in time and space complexity.

```python
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
    return arr

lst = [1, 12, 0, 2, 2, 6]
bubble_sort(lst)
```

Time Complexity: O(n²)

- Worst case: All elements are out of order (e.g., [5, 4, 3, 2, 1]).
For n elements, it requires n*(n-1)/2 comparisons.

Space Complexity: O(1)

- Sorts "in place" without creating copies of the array.

```python
def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    return merge(left, right)

def merge(left, right):
    merged = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            merged.append(left[i])
            i += 1
        else:
            merged.append(right[j])
            j += 1
    merged += left[i:]
    merged += right[j:]
    return merged

lst = [1, 12, 0, 2, 2, 6]
merge_sort(lst)
```

Time Complexity: O(n log n)

- Splitting the array into halves takes log n steps, and merging at each level takes O(n) time.

Space Complexity: O(n)

- Requires temporary arrays for merging (not strictly in-place).
