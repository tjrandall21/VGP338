### Part 4 - Bitwise Operations

```python
# Bitwise Operators
p = 6  # Binary: 110
q = 3  # Binary: 011

print(p & q)  # Bitwise AND: 110 & 011 = 010 (2 in decimal)
print(p | q)  # Bitwise OR:  110 | 011 = 111 (7 in decimal)
print(p ^ q)  # Bitwise XOR: 110 ^ 011 = 101 (5 in decimal)
print(~p)     # Bitwise NOT: ~110 = -111 (-7 in decimal)
print(p << 1) # Left shift: 110 << 1 = 1100 (12 in decimal)
print(p >> 1) # Right shift: 110 >> 1 = 011 (3 in decimal)
```

#### **`list` Big-O Summary**
| Operation               | Example                  | Big-O      | Explanation                          |
|-------------------------|--------------------------|------------|--------------------------------------|
| **Index Access**        | `lst[2]`                 | O(1)       | Direct pointer jump.                 |
| **Search**              | `"x" in lst`             | O(n)       | Worst-case: check all pointers.      |
| **Append**              | `lst.append(x)`          | O(1)*      | Usually instant, rare resizing.      |
| **Insert/Delete (End)** | `lst.pop()`              | O(1)       | No shifting needed.                  |
| **Insert/Delete (Mid)** | `lst.insert(3, x)`       | O(n)       | Pointers shifted left/right.         |
| **Length Check**        | `len(lst)`               | O(1)       | Pre-stored size, no counting.        |

### Part 7 - Sets

```python
# Sets
unique_numbers = {1, 2, 3, 3, 2}
a = [1, 1, 1, 2, 2, 4, 5, 5, 6] # list
b = set(a) # to set
b

# set operations
set_a = {1, 2, 3}
set_b = {3, 4, 5}
print(set_a | set_b)  # Union: {1, 2, 3, 4, 5}
print(set_a & set_b)  # Intersection: {3}
print(set_a - set_b)  # Difference: {1, 2}
print(set_a ^ set_b)  # Symmetric Difference: {1, 2, 4, 5}
```

| **Common Operations**              | **Average Time Complexity** | **Worst-Case Time Complexity** |
|----------------------------|-----------------------------|---------------------------------|
| Membership Test (`x in s`) | O(1)                        | O(n)                           |
| Add/Remove (`s.add(x)`, `s.remove(x)`, `s.discard(x)`) | O(1)                        | O(n)                           |
| Length (`len(s)`)          | O(1)                        | O(1)                           |
| Iteration (`for x in s`)   | O(n)                        | O(n)                           |

### Part 8 - Dictionaries

```python
# Dictionaries
storeSections = {"apples": 3, "bread": 1, "butter": 2, "pasta": 4}


# Iterating Through a Dictionary:
for item in storeSections:
    section = storeSections[item]
    print(f"{item} is in section {section}")

del storeSections["eggs"]
```

```python
from collections import defaultdict

# Provides a default value for nonexistent keys

groups = defaultdict(list) # default value would be an empty list []
words = ["apple", "banana", "apricot", "blueberry", "cherry"]

for word in words:
    groups[word[0]].append(word)

print(groups)
```

| Operation   | Example                 | Complexity | Explanation                              |  
|-------------|-------------------------|------------|------------------------------------------|  
| **Insert**  | `d["Alice"] = 5551234`  | O(1)       | Hash → Index → Direct write.             |  
| **Lookup**  | `d["Alice"]`            | O(1)       | Hash → Index → Direct read.              |  
| **Delete**  | `del d["Alice"]`        | O(1)       | Hash → Index → Remove entry.             |  
| **Iterate** | `for k in d:`           | O(n)       | Must scan all slots (even empty ones).   |  


| **Structure** | **Mutability** | **Memory**     | **Use Case**                   |  
|---------------|----------------|----------------|--------------------------------|  
| **List**      | Mutable        | High (pointers) | Dynamic, mixed-type data.     |  
| **Tuple**     | Immutable      | Low            | Fixed data.                    |  
| **String**    | Immutable      | Low            | Text processing.               |  
| **Array**     | Mutable        | Very Low       | Numeric efficiency.            |  

### Part 9 - Error Handling and Debugging


```python
# Syntax Error

print("this will work")
```

```python
# Runtime errors occur during execution, e.g., division by zero.
x = 0
try:
    y = 1 / x
except ZeroDivisionError as e:
    print(f"Runtime Error: {e}")  # Output: Runtime Error: division by zero

# Correct Code:
x = 1
y = 1 / x
```

```python
# Logical Errors
price = 1000
gst = 10  # 10 percent
total_cost = price * gst  # Incorrect logic
print(f"Logical Error Example: {total_cost}")  # Output: 10000 (incorrect)

# Correct Code:
gst = 0.1  # 10 percent
total_cost = price + (price * gst)  # Correct logic
print(f"Correct Calculation: {total_cost}")  # Output: 1100.0
```

```python
try:
    result = 10 / 0
except ZeroDivisionError as e:
    print("Error:", e)
finally:
    print("Execution completed.")
```
