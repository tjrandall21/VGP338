```python
STOPWORDS = {"the", "is", "a", "an", "and", "in", "to", "of", "it", "for", "on"}
PUNCTUATION = '.,!?;:"'
```

```python
import re
def clean_and_tokenize(text):
    if not isinstance(text, str):
        raise TypeError("clean_and_tokenize: argument is not a string")
    else:
        text = text.lower()
        text = text.translate(str.maketrans("","",PUNCTUATION))
        word_list = re.split(r'\s',text)
        return word_list
```

```python
clean_and_tokenize("John, Joseph?, Jackson!")
```

```python
import collections
def build_vocabulary(corpus, stopwords, min_freq=2):
    word_list = []
    for sentence in corpus:
        try:
            word_list += clean_and_tokenize(sentence)
        except TypeError:
            print("Warning: corpus element is not a string")
    word_list = [word for word in word_list if not word in stopwords]
    word_counts = dict(collections.Counter(word_list))
    unique_id = 1
    words_with_id = {}        
    for key, value in word_counts.items():
        if value >= min_freq:
            words_with_id[key] = unique_id
            unique_id += 1
    return words_with_id
```

```python
sample_corpus = [
    "Machine learning enables computers to learn from experience.",
    "Deep learning is a subset of machine learning.",
    "Artificial intelligence is transforming the modern world.",
    None,  # Should be caught by try-except without crashing
    "Learning algorithms require data and compute power.",
    #"transforming, algorithms, intelligence, modern, machine"
]

build_vocabulary(sample_corpus, STOPWORDS)
```

```python
class DataSample:
    
    def __init__(self, sample_id, label) -> None:
        self.sample_id = sample_id
        self.label = label
    def summary(self):
        return f"Sample #{self.sample_id} - Label: {self.label}"

class TextSample(DataSample):
    def __init__(self, sample_id, label, text) -> None:
        super().__init__(sample_id, label)
        self.text = text
    def word_count(self):
        words = re.split(r'\s',self.text)
        return len(words)
    def summary(self):
            return f"TextSample #{self.sample_id} ({self.word_count()} words) - Label: {self.label}"
```
